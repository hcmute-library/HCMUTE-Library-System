## Đọc data từ SQL Server

```python
query_phieumuon = """SELECT PMS.ID_phieu_muon, PMS.ID_ban_doc, BD.ID_lop, ID_khoa, BD.ID_nhom_nghanh_nghe, BD.ID_nien_khoa, PMS.Ngay_muon
                        FROM DIM_Phieu_muon_sach PMS
                                JOIN DIM_Ban_doc BD ON PMS.ID_ban_doc = BD.ID_ban_doc
                                JOIN DIM_Lop L ON BD.ID_Lop = L.ID_lop 
                        WHERE Ngay_muon >= 20040101 AND Ngay_muon < 20050101"""
df_phieumuon = pd.read_sql(query_phieumuon, conn_dwh_library)
print(df_phieumuon)
```

# Xử lý code

## tổng số lần mượn sách 

```python
# Tính số lượng mượn sách
so_luong_muon_sach = df_phieumuon.groupby(['ID_lop', 'ID_khoa', 'ID_nhom_nghanh_nghe', 'ID_nien_khoa', 'Ngay_muon'])['ID_ban_doc'].count().reset_index()
so_luong_muon_sach = so_luong_muon_sach.rename(columns={'ID_ban_doc': 'So_luong_muon_sach'})
print(so_luong_muon_sach)
```

## lượng sinh viên theo lớp

```python
# Tính số lượng sinh viên theo lớp
so_luong_sinh_vien_lop = df_phieumuon.groupby('ID_lop')['ID_ban_doc'].nunique().reset_index()
so_luong_sinh_vien_lop = so_luong_sinh_vien_lop.rename(columns={'ID_ban_doc': 'So_luong_sinh_vien_lop'})
print(so_luong_sinh_vien_lop)
```

## lượng sinh viên theo khoa

```python
# Tính số lượng sinh viên theo khoa
so_luong_sinh_vien_khoa = df_phieumuon.groupby('ID_khoa')['ID_ban_doc'].nunique().reset_index()
so_luong_sinh_vien_khoa = so_luong_sinh_vien_khoa.rename(columns={'ID_ban_doc': 'So_luong_sinh_vien_khoa'})
print(so_luong_sinh_vien_khoa)
```

## lượng sinh viên theo ngành

```python
# Tính số lượng sinh viên theo ngành
so_luong_sinh_vien_nghanh = df_phieumuon.groupby('ID_nhom_nghanh_nghe')['ID_ban_doc'].nunique().reset_index()
so_luong_sinh_vien_nghanh = so_luong_sinh_vien_nghanh.rename(columns={'ID_ban_doc': 'So_luong_sinh_vien_nghanh'})
print(so_luong_sinh_vien_nghanh)
```

## lượng sinh viên theo niên khóa

```python
# Tính số lượng sinh viên theo niên khóa
so_luong_sinh_vien_nien_khoa = df_phieumuon.groupby('ID_nien_khoa')['ID_ban_doc'].nunique().reset_index()
so_luong_sinh_vien_nien_khoa = so_luong_sinh_vien_nien_khoa.rename(columns={'ID_ban_doc': 'So_luong_sinh_vien_nien_khoa'})
print(so_luong_sinh_vien_nien_khoa)
```

## lượng sinh viên theo ngày

```python
# Tính số lượng sinh viên theo ngày
so_luong_sinh_vien_ngay = df_phieumuon.groupby('Ngay_muon')['ID_ban_doc'].nunique().reset_index()
so_luong_sinh_vien_ngay = so_luong_sinh_vien_ngay.rename(columns={'ID_ban_doc': 'So_luong_sinh_vien_ngay'})
print(so_luong_sinh_vien_ngay)
```

## Gộp lại các kết quả

```python
# Merge 'so_luong_muon_sach' và 'so_luong_sinh_vien_lop' theo 'ID_lop'
result = pd.merge(so_luong_muon_sach, so_luong_sinh_vien_lop, on='ID_lop', how='left')
result = pd.merge(result, so_luong_sinh_vien_khoa, on='ID_khoa', how='left')
result = pd.merge(result, so_luong_sinh_vien_nghanh, on='ID_nhom_nghanh_nghe', how='left')
result = pd.merge(result, so_luong_sinh_vien_nien_khoa, on='ID_nien_khoa', how='left')
result = pd.merge(result, so_luong_sinh_vien_ngay, on='Ngay_muon', how='left')
# Kiểm tra kết quả
print(result)
```
