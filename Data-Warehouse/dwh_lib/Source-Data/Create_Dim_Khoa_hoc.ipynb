```python
import os
import pandas as pd
```

```python
start_year = 1970 # Khởi tạo năm bắt đầu
end_year = 2050  # Kết thúc ở năm 2024
data = {"ID khóa học": [], "Tên khóa học": []}

for i in range(1970, 2000):
    data["ID khóa học"].append(str(i - 1970 + 70))  # Sử dụng số đơn giản cho ID
    data["Tên khóa học"].append(f"{i} - {i + 4}")

for j in range(2000, end_year + 1):
    data["ID khóa học"].append(f"K{j - 2000}")  # Bắt đầu từ K0
    data["Tên khóa học"].append(f"{j} - {j + 4}")

df_Dim_Khoa = pd.DataFrame(data)
print(df_Dim_Khoa)
```

```python
file_path = './Data_Dim_Khoa.csv'
if os.path.exists(file_path):
    os.remove(file_path)
df_Dim_Khoa.to_csv(file_path, index=False, encoding='utf-8-sig')
```
