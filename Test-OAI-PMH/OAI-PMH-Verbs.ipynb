## 1. **Identify**
Cung cấp thông tin về kho lưu trữ, chẳng hạn như tên, mô tả, phiên bản OAI-PMH mà nó hỗ trợ.

```python
url = "http://cds.cern.ch/oai2d?verb=Identify"
response = requests.get(url)
print(response.text)
```

## 2. ListMetadataFormats
Trả về danh sách các định dạng siêu dữ liệu mà kho lưu trữ hỗ trợ (ví dụ: Dublin Core, MARCXML, METS).

```python
url = "http://cds.cern.ch/oai2d?verb=ListMetadataFormats"
response = requests.get(url)
print(response.text)
```

## 3. ListSets
Liệt kê các tập hợp con (sets) trong kho lưu trữ, nếu có. Đây là các phân loại khác nhau của siêu dữ liệu.

```python
url = "http://cds.cern.ch/oai2d?verb=ListSets"
response = requests.get(url)
print(response.text)
```

## 4. GetRecord
Lấy một bản ghi siêu dữ liệu cụ thể từ kho lưu trữ bằng cách sử dụng một định danh duy nhất.

### 4.1 GetRecord (Dublin Core)

```python
url = "http://cds.cern.ch/oai2d?verb=GetRecord&identifier=oai:cds.cern.ch:2910744&metadataPrefix=oai_dc"
response = requests.get(url)
print(response.text)
```

### 4.2 GetRecord (MARCXML)
